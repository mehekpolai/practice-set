## Solve using pandas

```python
# what is the avg math score
print(df["math score"].mean())
```

```python
# highest math score
print(df["math score"].max())
```

```python
# math score above 80
len(df[df["math score"]>80])
```

```python
# avg reading score by gender
df.groupby("gender")["reading score"].mean()
```

## creating a simple visualization
