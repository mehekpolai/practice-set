## Data Cleaning

```python
# Find missing values:
df.isnull().sum()
```

```python
#Replace missing directors:
df['director'] = df['director'].fillna('Unknown')
```

```python
#Replace missing countries:
df['country'] = df['country'].fillna('Unknown')
```

```python
#Remove duplicate records:
df = df.drop_duplicates()
```

```python
#Convert date_added into datetime:
df['date_added'] = pd.to_datetime(
    df['date_added'].str.strip(),
    errors='coerce'
)
```

## Data visualization using Matplotlib and Seaborn

```python
# Movies vs TV Shows
sns.countplot(data=df, x='type')

plt.title('Movies vs TV Shows')
plt.xlabel('Content Type')
plt.ylabel('Count')

plt.show()
```

```python
# Top 10 countries
top_countries = (
    df['country']
    .str.split(',')
    .explode()
    .str.strip()
    .value_counts()
    .head(10)
)

top_countries.plot(
    kind='bar',
    color='skyblue',
    figsize=(10,5)
)

plt.title('Top 10 Countries by Netflix Titles')
plt.xlabel('Country')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)
plt.show()
```

```python
# Content released each year
yearly = df['release_year'].value_counts().sort_index()

plt.figure(figsize=(12,5))

plt.plot(yearly.index, yearly.values)

plt.title('Netflix Content Released Each Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')

plt.show()
```

```python
# Content rating distribution
plt.figure(figsize=(12,5))

sns.countplot(
    data=df,
    x='rating',
    order=df['rating'].value_counts().index
)

plt.title('Netflix Content Ratings')
plt.xticks(rotation=45)

plt.show()
```
